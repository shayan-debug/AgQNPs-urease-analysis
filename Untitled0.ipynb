```python
from rdkit.Chem import Descriptors

def lipinski(mol):
    mw = Descriptors.MolWt(mol)
    logp = Descriptors.MolLogP(mol)
    hbd = Descriptors.NumHDonors(mol)
    hba = Descriptors.NumHAcceptors(mol)

    print(f"MW:   {mw:.1f}  (باید < 500)")
    print(f"LogP: {logp:.2f} (باید < 5)")
    print(f"HBD:  {hbd}     (باید < 5)")
    print(f"HBA:  {hba}     (باید < 10)")
    print()
    passes = mw<500 and logp<5 and hbd<5 and hba<10
    print("✅ Lipinski pass" if passes else "❌ Lipinski fail")

lipinski(quercetin)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# داده‌های DFT از Table 3 — Shayan Asadi et al., Scientific Reports 2025
states = ['Singlet', 'Doublet']
interaction_energies = [-53.79, -50.77]  # kcal/mol

colors = ['#2196F3', '#4CAF50']

fig, ax = plt.subplots(figsize=(7, 4))
bars = ax.bar(states, interaction_energies, color=colors, width=0.4, edgecolor='white')

ax.set_ylabel('Interaction Energy (kcal/mol)')
ax.set_title('Quercetin–Ag Sheet Interaction Energy\nShayan Asadi et al., Scientific Reports 2025')
ax.set_ylim(-65, 0)

for bar, val in zip(bars, interaction_energies):
    ax.text(bar.get_x() + bar.get_width()/2, val - 1.5,
            f'{val} kcal/mol', ha='center', va='top', fontsize=11, color='white', fontweight='bold')

plt.tight_layout()
plt.savefig('dft_interaction_energy.png', dpi=150)
plt.show()
```

```python
# IC50 data از مقاله — Shayan Asadi et al., Scientific Reports 2025
samples = ['Quercetin\n(pure)', 'AgNPs\n(alone)', 'Ag@QNPs\n(conjugated)']
ic50 = [5610, None, 22.4]  # µg/mL — AgNPs alone IC50 نداشت

fig, ax = plt.subplots(figsize=(7, 4))
colors = ['#FF9800', '#9E9E9E', '#2196F3']

vals = [5610, 0, 22.4]
bars = ax.bar(samples, vals, color=colors, width=0.4, edgecolor='white')

ax.set_ylabel('IC₅₀ (µg/mL) — lower is better')
ax.set_title('Urease Inhibitory Activity\nShayan Asadi et al., Scientific Reports 2025')

ax.text(1, 50, 'No IC₅₀\n(inactive)', ha='center', fontsize=9, color='gray')
ax.text(0, 5800, '5610', ha='center', fontsize=10, fontweight='bold')
ax.text(2, 120, '22.4', ha='center', fontsize=10, fontweight='bold', color='white')

plt.tight_layout()
plt.savefig('ic50_comparison.png', dpi=150)
plt.show()
```

```python
import numpy as np

samples = ['Quercetin\n(pure)', 'Ag@QNPs\n(conjugated)']
ic50 = [5610, 22.4]
colors = ['#FF9800', '#2196F3']

fig, ax = plt.subplots(figsize=(6, 4))
bars = ax.bar(samples, ic50, color=colors, width=0.4)

ax.set_yscale('log')
ax.set_ylabel('IC₅₀ (µg/mL) — log scale — lower is better')
ax.set_title('Urease Inhibitory Activity\nShayan Asadi et al., Scientific Reports 2025')

for bar, val in zip(bars, ic50):
    ax.text(bar.get_x() + bar.get_width()/2, val * 1.3,
            f'{val}', ha='center', fontsize=11, fontweight='bold')

ax.text(0.5, 0.5, '250× more potent →', transform=ax.transAxes,
        ha='center', fontsize=9, color='gray', style='italic')

plt.tight_layout()
plt.savefig('ic50_comparison.png', dpi=150)
plt.show()
```
